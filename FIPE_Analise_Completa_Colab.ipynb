```
# Selecionar as colunas especificadas
df_Toyota_dep5 = df_Toyota_clean[['codigoFipe', 'marca', 'modelo', 'anoModelo', 'anoReferencia', 'valor']].copy()

# Ensure 'valor' column is of type string before applying string methods
df_Toyota_dep5['valor'] = df_Toyota_dep5['valor'].astype(str)

# Remover símbolo de moeda e converter valor para tipo numérico float, se necessário
df_Toyota_dep5['valor'] = df_Toyota_dep5['valor'].str.replace('[R$]', '', regex=True).str.replace(',', '').astype(float)

# Obter o valor de 2022 para cada codigoFipe
df_2022 = df_Toyota_dep5[df_Toyota_dep5['anoReferencia'] == 2022][['codigoFipe', 'valor']].rename(columns={'valor': 'valor_2022'})

# Mesclar para adicionar o valor de 2022 ao DataFrame original
df_Toyota_dep5 = df_Toyota_dep5.merge(df_2022, on='codigoFipe', how='left')

# Calcular a desvalorização total para cada codigoFipe (com ajuste para valores positivos)
# Instead of using last() and first(), use iloc[-1] and iloc[0] to get the last and first values
df_Toyota_dep5['desv/valorizacao'] = (
    df_Toyota_dep5.groupby('codigoFipe')['valor']
    .transform(lambda x: ((x.iloc[-1] - x.iloc[0]) / x.iloc[0]) * 100)  # Use iloc to access values by position
    .round(2)
)

# Filtrar para manter apenas o valor de 2022 e as colunas finais
df_Toyota_dep5 = df_Toyota_dep5[df_Toyota_dep5['anoReferencia'] == 2022][['codigoFipe', 'marca', 'modelo', 'anoModelo', 'valor_2022', 'desv/valorizacao']]
df_Toyota_dep5 = df_Toyota_dep5.rename(columns={'valor_2022': 'valor'})

# Exibir o DataFrame resultante
print(df_Toyota_dep5)
```
